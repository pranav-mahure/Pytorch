```python
## Tensor datatype
#float32 tensor
float_32_tensor = t.tensor([3.0,6.0,9.0],
                           dtype=t.float32,
                           device=None,  # cuda or cpu or what device is your tensor on
                            requires_grad=False)  # Whether or not to track gradient with these tensor operation
float_32_tensor                           
```

* If dtype= written None it will still gives float32 as it's default

* Float32 means a number will take 32 bits of storage in memory of computer float16 will take 16 bit of memory in computer

* ' Tensor datatype is one of the three big issues in Python And deep learning basically you will tackle as an error while writing the code Remember this'
1. tensor not right dtatype
2. tensor not right shape
3. tensors not on the right device (cpu or gpu)



```python
float_32_tensor.dtype
```

```python
float_16_tensor = float_32_tensor.type(t.float16)
float_16_tensor
```

```python
float_16_tensor.dtype
```
