### Matrix Multiplication
There are multiple ways to perform matrix multiplication in PyTorch. Here are a few common methods:

1. Using the `@` operator (Python 3.5+):
   ```python
   result = tensor_a @ tensor_b
   ```
2. Using the `torch.matmul()` function:
   ```python

* aslo there are 2 rules for matrix multiplication
1. The inner dimensions must match
* `(3,2) @ (3,2) ` won't work
* `(2,3) @ (3,2)` will work
* `(3,2) @ (2,3)` will work

2. The resulting matrix will have the shape of the outer dimensions
* `(2,3) @ (3,2)` -> `(2,2)` 

````python
## first let's do dot multiplication
print(tensor, "*" ,tensor)
print(f"Equals :- {tensor * tensor}")
````

````python
#matrix
torch.matmul(tensor,tensor)  ## internally faster
````

To fix tensor shape issue we can shape our tensor using transpose.
A transpose is swapping the dimensions of a given tensor.
* For example if we have a tensor of shape (3,2) and we want to perform matrix multiplication with a tensor of shape (2,3) we can transpose the first tensor to make it (2,3) so that the inner dimensions match and we can perform matrix multiplication.
* result = torch.matmul(tensor_a, tensor_b)

Reshaping, stacking, squeezing and unsqueezing
Often times you'll want to reshape or change the dimensions of your tensors without actually changing the values inside them.

To do so, some popular methods are:



Why do any of these?

Because deep learning models (neural networks) are all about manipulating tensors in some way. And because of the rules of matrix multiplication, if you've got shape mismatches, you'll run into errors. These methods help you make sure the right elements of your tensors are mixing with the right elements of other tensors.

````python
## let's create tensor

x = torch.arange(1.,10.)
x , x.shape
````

````python
## add an extra dimension
x_reshaped =  x.reshape(3,3)
x_reshaped , x_reshaped.shape
````

````python
z = x.view(1,9)
z , z.shape
````

````python
# changing z changes x (because a view off a tensor shares the same memory as the original)
z[:, 0] = 5
z , x
````

````python
## stack tensors on top of each other

x_stacked = torch.stack([x,x,x,x], dim =0)
x_stacked, x_stacked.shape
````

````python
## squeeze
x_reshaped , x_reshaped.shape
````

````python
x_reshaped.squeeze() , x_reshaped.squeeze().shape ## removes single dimension that is if shape would have been (1,3) it would had been (3) Try out
````

````python
x_reshaped.squeeze() , x_reshaped.squeeze().shape ## removes single dimension that is if shape would have been (1,3) it would had been 
x_squeezed = x_reshaped.squeeze()
````

````python
# torch.unsqueeze() -  adds a single dimension to a target tensor at a specific dim 
X_unsqueezed = x_squeezed.unsqueeze(dim=1)
X_unsqueezed , X_unsqueezed.shape
````

## reproducibility in pytorch

````python
## to reduce the randomness in neural networks and pytorch comes the concept of a random seed
# Essentially what the random seed does is "flavour" the randomness

random_tensor_A = torch.rand(3, 4)
random_tensor_B  = torch.rand(3 ,4)
````

````python
print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)
````

````python
## let's make some random but reproducibale tensor
import torch

# set the random seed
RANDOM_SEED = 42
torch.manual_seed(RANDOM_SEED)
random_tensor_C = torch.rand(3, 4)

torch.manual_seed(RANDOM_SEED)
random_tensor_D = torch.rand(3, 4)

print(random_tensor_C)
print(random_tensor_D)
print(random_tensor_C == random_tensor_D)
````
